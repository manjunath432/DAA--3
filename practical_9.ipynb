```python
def prim(graph):
    n = len(graph)
    selected = [False] * n
    selected[0] = True

    print("Edge  Weight")

    for _ in range(n - 1):
        min_weight = float('inf')
        x = y = -1

        for i in range(n):
            if selected[i]:
                for j in range(n):
                    if not selected[j] and graph[i][j] != 0:
                        if graph[i][j] < min_weight:
                            min_weight = graph[i][j]
                            x, y = i, j

        print(x, "-", y, " ", min_weight)
        selected[y] = True


# Graph represented as an adjacency matrix
graph = [
    [0, 2, 3, 0, 0],
    [2, 0, 1, 4, 0],
    [3, 1, 0, 5, 6],
    [0, 4, 5, 0, 7],
    [0, 0, 6, 7, 0]
]

prim(graph)
```
